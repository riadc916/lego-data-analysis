### Aggregate Data with the Python .agg() Function

Let's work out the number of different themes shipped by year. This means we have to count the number of unique theme_ids per calendar year.

````python
themes_by_year=sets_df.groupby('year').agg({'theme_id':pd.Series.nunique})
````

````python
themes_by_year.rename(columns={'theme_id':'nr_themes'},inplace=True)
````

**Challenge**: Plot the number of themes released by year on a line chart. Only include the full calendar years (i.e., exclude 2020 and 2021).

````python
plt.figure(figsize=(16,10))
plt.plot(themes_by_year.index[:-2],themes_by_year.values[:-2],color='orange',linewidth=3)
plt.title('Number of LEGO Themes Published Over Time',fontsize=18)
plt.xlabel('Year',fontsize=14)
plt.ylabel('Number of Themes',fontsize=14)
plt.show()
````

### Line Charts with Two Seperate Axes

````python
plt.figure(figsize=(16,10))
ax1=plt.gca()
ax2=ax1.twinx()

ax1.plot(themes_by_year.index[:-2], themes_by_year['nr_themes'][:-2], color='orange', linewidth=3)
ax1.set_xlabel('Year',fontsize=14)
ax1.set_ylabel('Number of Themes',color='orange',fontsize=14)
ax1.tick_params(axis='y',labelcolor='orange')

ax2.plot(sets_by_year.index[:-2], sets_by_year.values[:-2], color='blue', linewidth=3)
ax2.set_ylabel('Number of sets',color='blue',fontsize=14)
ax2.tick_params(axis='y',labelcolor='blue')

plt.title('LEGO Themes vs Number of Sets Over Time', fontsize=18)

plt.show()
````

**Challenge**: Use the <code>.groupby()</code> and <code>.agg()</code> function together to figure out the average number of parts per set. How many parts did the average LEGO set released in 1954 compared to say, 2017?

````python
parts_by_year=sets_df.groupby('year').agg({'num_parts':pd.Series.mean})
parts_by_year.rename(columns={'num_parts':'avg_parts'},inplace=True)
````

````python
print(f"Average number of parts per set in 1954 : {parts_by_year.loc[1954]}")
print(f"Average number of parts per set in 2017 : {parts_by_year.loc[2017]}")
````

### Scatter Plots in Matplotlib

### Merging (i.e., Combining) DataFrames based on a Key


````python
set_theme_count=pd.DataFrame({
    'id':set_theme_count.index,
    'set_count':set_theme_count.values
})
````

````python
merge_df=pd.merge(set_theme_count,theme,on='id')
merge_df.head(3)
````

````python
plt.figure(figsize=(14, 8))
plt.xticks(fontsize=14, rotation=45)
plt.yticks(fontsize=14)
plt.ylabel('Nr of Sets', fontsize=14)
plt.xlabel('Theme Name', fontsize=14)

plt.bar(merge_df.name[:10], merge_df.set_count[:10])
plt.show()
````
